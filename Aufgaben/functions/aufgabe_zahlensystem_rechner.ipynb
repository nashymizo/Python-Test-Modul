# Umrechnerfunktion

- HEX Zahl soll als String behandelt werden
- Dezimalzahl soll als INTEGER behandelt werden
- Dualzahl soll als String behandelt werden

1. Dual zu Dezimal        Additionsverfahren
2. Dezimal zu Dual        Divisionsverfahren
3. Hex zu Dezimal         Additionsverfahren
4. Dezimal zu Hex         Divisionsverfahren
5. Dual zu Hex            dual -> dezimal -> hex
6. Hex zu Dual            hex -> dezimal -> dual


#3  = 3  = 0011
#9  = 9  = 1001
#A  = 10 = 1010

Nicht die eingebauten Funktionen benutzen.

Es sollen eigene Funktionen geschaffen werden.


```python
def hauptrechner(callback, zahl):
    hexadezi, normal, binaer = callback(zahl)
    print(f"#{hexadezi} = {normal} = {binaer}")
    return normal, binaer, hexadezi



def dual_dezi_hex(zahl):

    dezi = dual_dezi(zahl)
    hex = dezi_hex(dezi)

    return hex, dezi, zahl



def hex_dezi_dual(zahl):

    dezi = hex_dezi(zahl)
    dual = dezi_dual(dezi)

    return zahl, dezi, dual




'''
DUAL TO DEZI
'''

def dual_dezi(zahl):
    dual = 2
    potenz = len(zahl) - 1
    dualzahl= 0
    for zahl in zahl:
        if zahl == "1":
            dualzahl = dualzahl + (dual ** potenz)
            potenz -= 1
        else:
            potenz -= 1
    return dualzahl

print(dual_dezi('1111'))

# def dual_dezi_v2(zahl):
#
#     dezimal_zahl = int(zahl, 2)
#
#     return dezimal_zahl
#
# print(dual_dezi_v2("1111"))

'''
DEZI TO DUAL

 ODER MIT:
                    ergebnis += str(rest)
                    ergebnis[::-1]      # Kehrt den String um

'''

def dezi_dual(zahl):

    ergebnis = ""

    while zahl > 0:
        rest = zahl % 2
        ergebnis = (str(rest) + ergebnis)

        zahl //= 2
    return ergebnis

print(dezi_dual(198))

# def dezi_dual_v2(zahl):
#     return bin(zahl)[2:]  # bin() wandelt um, [2:] schneidet das '0b'-Praefix ab
#
# print(dezi_dual_v2(198))

'''
HEX TO DEZI
'''

def hex_dezi(zahl):

    hex_dict = {"A": 10, "B" : 11, "C" : 12, "D" : 13, "E" : 14, "F" : 15}

    potenz = len(zahl) - 1
    hex_multi = 16
    ergebnis = 0

    for wert in zahl:
        if wert.upper() in hex_dict:
            faktor = (hex_dict[wert.upper()])
            ergebnis = ergebnis + faktor * (hex_multi ** potenz)
            potenz -= 1

        else:
            faktor = int(wert)
            ergebnis = ergebnis + faktor * (hex_multi ** potenz)
            potenz -= 1

    return ergebnis

print(hex_dezi("2f3"))

# def hex_dezi_v2(zahl):
#     ergebnis = int(zahl, 16)
#     return ergebnis
#
# print(hex_dezi_v2("2f3"))

'''
DEZI TO HEX
'''

def dezi_hex(zahl):

    hex_dict = {10: "A", 11: "B", 12: "C", 13: "D", 14: "E", 15: "F"}
    ergebnis = ""
    rest = zahl

    while rest > 0:
        if rest % 16 < 10:
            ergebnis = ergebnis + str(rest % 16)
            rest = rest // 16

        else:
            teil_ergebnis = rest % 16
            ergebnis_dict = hex_dict[teil_ergebnis]
            ergebnis = ergebnis + str(ergebnis_dict)
            rest = rest // 16


    return ergebnis[::-1]

print(dezi_hex(198))

# def dezi_hex_v2(zahl):
#     return hex(zahl)[2:].upper()
#
# print(dezi_hex_v2(198))  # Ausgabe: C6

hauptrechner(dual_dezi_hex, "11100111")

hauptrechner(hex_dezi_dual, "E7")


```
