Ein simpler Cache, der das Ergebnis einer Funktion zwischenspeichert.

Tipp;
- im Cache muss man die Daten wiederfinden
    - Speicherung der Daten in einem
    - eigener Schlüssel bilden -> Hashing -> MD5
        - Funktionsname
        - *args, **kwargs
- mit Funktionsnamen und den Parametern einen eindeutigen schlüssel bilden

Diese Technik nennt man Memoization

```python
# # Beispiel wie der Cache aussehen könnte
# cache = {
#     "abcdef123": "A",
#     "213412fda": "B",
#     "2352daaad": "C"
# }
```

```python
import time
import hashlib

def simple_cache(func):
    cache = {}
    def wrapper(*args, **kwargs):

        name_function = func.__name__

        kwargs_sorted = sorted(kwargs.items())

        test_string =f"{name_function}_{args}_{kwargs_sorted}"

        test_string_bytes = test_string.encode("utf-8")

        test_string_hash_object = hashlib.md5(test_string_bytes)

        test_hash = test_string_hash_object.hexdigest()

        if test_hash in cache:
            return cache[test_hash]

        else:
            result = func(*args, **kwargs)

            cache[test_hash] = result

            return result
    return wrapper





@simple_cache
def get_userdata(id):
    time.sleep(2) # Simulierung einer langsamen abfrage
    return {"id":id, "name":"user_"+str(id), "age": 20}

@simple_cache
def addiere(x, y):
    print("addiere aufgerufen")
    time.sleep(2)
    return x + y


print("dauert 2 sekunden", addiere(x=5, y=3))  # dauert 2 sekunden
print("kommt sofort", addiere(y=3, x=5))  # kommt sofort
print("dauert 2 sekunden", addiere(x=10, y=20))  # dauert 2 sekunden
print("kommt sofort", addiere(y=20, x=10))  # kommt sofort
print("kommt sofort", addiere(x=5, y=3))  # kommt sofort
```
