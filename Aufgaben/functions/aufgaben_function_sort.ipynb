- Sortiere die Liste der Bücher nach Erscheinungsjahr aufsteigend.
- Sortiere die Liste der Bücher nach der Länge des Titels.
- Sortiere die Liste der Bücher nach Autor alphabetisch aufsteigend, das Erscheinungsjahr absteigend.

```python
buecher = [
    {"titel": "Der Prozess", "autor": "Franz Kafka", "jahr": 1925},
    {"titel": "Faust", "autor": "Johann Wolfgang von Goethe", "jahr": 1808},
    {"titel": "Die Verwandlung", "autor": "Franz Kafka", "jahr": 1915},
    {"titel": "Die Physiker", "autor": "Friedrich Dürrenmatt", "jahr": 1962},
    {"titel": "Homo Faber", "autor": "Max Frisch", "jahr": 1957},
]
# SORTIERUNG NACH JAHR !

sort_year = sorted(buecher, key=lambda buch: buch["jahr"])
print(sort_year)

# SORTIERUNG NACH länge des buches

sort_len_title = sorted(buecher, key=lambda buch: len(buch["titel"]))
print(sort_len_title)

# SORTIERUNG NACH autor AUFSTEIGEND und jahr ABSTEIGEND


sort_test = sorted(buecher, key=lambda buch: (buch["autor"], -buch["jahr"]))

print(sort_test)
```

Sortiere die Liste nach dem Alter (zweites Element des Tupels).

```python
personen = [("Alice", 25), ("Bob", 30), ("Charlie", 20)]

personen_neu = sorted(personen, key=lambda person: person[1])

print(personen_neu)
```

Aufgabe: Sortiere die Liste nach den Noten (aufsteigend).

```python
students = [
    {"name": "Alice", "note": 2.5},
    {"name": "Bob", "note": 1.7},
    {"name": "Charlie", "note": 3.0}
]

students_neu = sorted(students, key=lambda student: student["note"])
print(students_neu)
```

Sortiere zuerst nach Alter, dann nach Note (aufsteigend).

```python
students2 = [
    {"name": "Alice", "note": 2.5, "alter": 25},
    {"name": "Bob", "note": 1.7, "alter": 25},
    {"name": "Charlie", "note": 3.0, "alter": 20}
]

students2_neu = sorted(students2, key=lambda student: (student["alter"], student["note"]))
print(students2_neu)
```
