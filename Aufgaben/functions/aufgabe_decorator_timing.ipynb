Schreibe einen Dekorator @timing, der misst, wie lange die Ausführung einer Funktion dauert.

Verwende ihn dann bei einer Funktion, die eine große Liste mit Zufallszahlen sortiert.

Die Ausgabe der gemessenen Zeit soll direkt im Wrapper() erfolgen.

Zusatzaufgabe:

Wer Bock hat kann ja mal die Zeitmessung in eine Datei schreiben

```python
import random
import time


test_list = []
for i in range(100):
    neue_zahl = random.randint(0,100000000)
    test_list.append(neue_zahl)

def timing(func):
    def wrapper(*args, **kwargs):

        start_time = time.time()

        result = func(*args, **kwargs)

        end_time = time.time()

        result_time = str(end_time - start_time)
        result_time = result_time.replace(".", ",")


        print(f"Die Dauer der Sortierung beträgt: {result_time} Sekunden")

        with open("decorator_timing.txt", "w", encoding="utf8") as f:
            f.write(f"Die Dauer der Sortierung beträgt: {result_time} Sekunden")

        return result
    return wrapper



@timing
def sortiere_liste(liste):
    new_list = sorted(liste)
    time.sleep(2) #dauer 2 sekunden
    return new_list


sortiere_liste(test_list)

```
