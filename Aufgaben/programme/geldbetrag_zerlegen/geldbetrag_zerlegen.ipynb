# Lösung 1

````python
def geld_zerlegen(betrag_euro: float):

    werte_cent = [
        50000,
        20000,
        10000,
        5000,
        2000,
        1000,
        500,
        200,
        100,
        50,
        20,
        10,
        5,
        2,
        1,
    ]

    #betrag in Cent umrechnen, um mit ganzen Zahlen zu arbeiten
    rest_cent = round(betrag_euro * 100)

    ergebnisse = []

    for wert in werte_cent:

        anzahl = rest_cent // wert

        #Restbetrag aktualisieren
        rest_cent = rest_cent % wert

        #rest in Euro umrechnen
        wert_euro = wert / 100

        #ohne Komma, wenn es eine ganze Zahl ist
        if wert_euro.is_integer():
            ergebnisse.append(f"{anzahl}x {int(wert_euro)} €")
        else:
            ergebnisse.append(f"{anzahl}x {wert_euro} €")

    #einString mit allen Ergebnissen zurückgeben, getrennt durch Leerzeichen
    return ergebnisse


# Testlauf
eingabe = 1123.88
ausgabe = geld_zerlegen(eingabe)

print(f"Eingabe: {eingabe}")
for item in ausgabe:
    print(item)
````
