Nur positive Eingaben summieren

Der Benutzer gibt solange Zahlen ein, bis er eine 0 eingibt. Es sollen nur die positiven Zahlen aufsummiert werden.

```python

def user_input():

    while True:
        try:

            zahl_abfrage = int(input("Bitte gib eine positive Zahl ein! (Mit 0 beendest du die Dauerschleife)"))

            return zahl_abfrage


        except Exception as e:

                print(e)



def positive_zahlen():

    ergebnis = 0

    while True:
        zahl = user_input()
        if zahl == 0:
            break

        elif zahl > 0:
            print(f"{zahl} (neu) + {ergebnis} (alt) = {zahl + ergebnis}")
            ergebnis = ergebnis + zahl

        elif zahl < 0:
            print("Die Zahl muss eine POSITIVE Zahl sein!!")


    return f"Das Ergebnis lautet: {ergebnis}"

    # while True:
    #     zahl = user_input()
    #     print(f"{zahl}(neu) + {ergebnis:02d}(alt) = {zahl + ergebnis}")
    #     ergebnis = ergebnis + zahl
    #     if zahl == 0:
    #         break
    # return f"Das Aufsummieren wurde beendet, dein Ergebnis lautet: {ergebnis}"


positive_zahlen()

```
