Erzeuge eine neue Liste:

- Werte ab 50 sollen "bestanden" ergeben
- Werte unter 50 sollen ""nicht bestanden" ergeben

```python
punkte = [35, 50, 72, 49]
```

```python
for zahl in punkte:
    if zahl < 50:
        print(f"{zahl} Punkte - Verkackt")
    else:
        print(f"{zahl} Punkte - Du hast bestanden")
```

```python
punkte = [35, 50, 72, 49]

ergebnisse = []
for zahl in punkte:
    if zahl >= 50:
        ergebnisse.append("bestanden")
    else:
        ergebnisse.append("nicht bestanden")
print(ergebnisse)
```

Eine verschachtelte Liste in lineare Liste wandeln (flatten)
Was bedeuten flatten?

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
matrix_neu = []

for reihe in matrix:
    for zahl in reihe:
        matrix_neu.append(zahl)

print(matrix_neu)
```

Wörter nach Länge filtern
Erzeuge eine Liste aller Wörter, die länger als 3 Buchstaben haben, und wandle sie in Großbuchstaben um.

```python
words = ["house", "dog", "international", "cup", "AI"]

words_gross = []

for items in words:
    if len(items) > 3:
        words_gross.append(items.upper())


print(words_gross)
```

# Sitzplatzanalyse

```python
sitzplan = [
    ["Jason", "Barry", "Mark", "Peter", "Andreas", "Maik"],
    ["John", "Jane", "Susi", "Sara", "Sven"],
    ["Max", "Moritz", "Mia", "Anna", "Markus"],
    ["Lukas", "Lena", "Joshua", "Luke", "Oliver"],
]

# Summe aller plätze
#menge = ???

# Die Summe der Plätze in einer Reihe
```

```python
sitze_overall = 0
sitzreihe_zaehler = 1

for reihe in sitzplan:
    sitze_reihe = 0

    for sitz in reihe:
        sitze_overall += 1
        sitze_reihe += 1

    print(f"In Reihe '{sitzreihe_zaehler}', sitzen '{sitze_reihe}' Personen")
    sitzreihe_zaehler += 1

print(f"----AUSVERKAUFT----")
print(f"Insgesamt wurden {sitze_overall} Tickets verkauft")


```
