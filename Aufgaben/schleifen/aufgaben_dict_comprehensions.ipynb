**Zeichenhäufigkeit. Tipp: Ein String kann man wie ein Array durchlaufen!**

````python
text = "Mississippi"
````

````python
anzahl = {}
for Buchstabe in text:
    if Buchstabe in anzahl:
        anzahl[Buchstabe] += 1
    else:
        anzahl[Buchstabe] = 1

print(anzahl)
````

**Häufigkeit zählen mit Dict Comprehension**

````python
list = ["Apfel", "Banane", "Apfel", "Kirsche", "Banane", "Apfel"]
````

````python
anzahl = {frucht: list.count(frucht) for frucht in list }
print(anzahl)
````

**ASCII-Tabelle**

`chr()` wandelt Zahl in Buchstabe, `ord()` gibt ASCII-Code.

````python
ascii_tabelle = {chr(zahl): zahl for zahl in range(ord('A'), ord('Z') + 1)}
print(ascii_tabelle)
````

Ausgabe:
```
{'A': 65, 'B': 66, ..., 'Z': 90}
```

**Preise inkl. Mehrwertsteuer**

````python
prices_netto = {"Apfel": 0.80, "Banane": 0.50, "Kirsche": 0.10}

for zeichen in prices_netto:
    print(f"{zeichen}, kostet: {prices_netto[zeichen]:.2f}€")
    mehrwertsteuer = prices_netto[zeichen] * 0.19
    endpreis = prices_netto[zeichen] + mehrwertsteuer
    print("mit Mehrwertsteuer:")
    print(f"{zeichen}, kostet: {endpreis:.2f}€")
    print(20 * "-")


````

````python
prices_netto = {"Apfel": 0.80, "Banane": 0.50, "Kirsche": 0.10}

prices_brutto = {Frucht: round(preis * 1.19, 2) for Frucht, preis in prices_netto.items()}

print(prices_brutto)
print(prices_brutto["Apfel"])
print(prices_brutto["Banane"])
print(prices_brutto["Kirsche"])
print(f"Der Preis für einen Apfel beträgt {prices_brutto['Apfel']}€")
````

````python
prices_netto = {"Apfel": 0.80, "Banane": 0.50, "Kirsche": 0.10}

# Wir erstellen ein neues Dictionary in einer Zeile
prices_brutto = {artikel: round(preis * 1.19, 2) for artikel, preis in prices_netto.items()}

print(prices_brutto)
# Ausgabe wäre: {'Apfel': 0.95, 'Banane': 0.59, 'Kirsche': 0.12}
````
